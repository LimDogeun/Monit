## 시도별 데이터 다운
- 우선 경기 다운

```python
import requests
import pandas as pd
import time
import os
import json
from tqdm import tqdm

ZCODE_DICT = {
    11: "서울특별시", 26: "부산광역시", 27: "대구광역시", 28: "인천광역시", 29: "광주광역시",
    30: "대전광역시", 31: "울산광역시", 36: "세종특별자치시", 41: "경기도", 43: "충청북도",
    44: "충청남도", 46: "전라남도", 47: "경상북도", 48: "경상남도", 50: "제주특별자치도",
    51: "강원특별자치도", 52: "전북특별자치도"
}

def normalize_region_name(region_name: str) -> str:
    return region_name.replace(" ", "").replace("특별자치시", "").replace("광역시", "").replace("도", "")

def get_total_count_by_zcode(api_key: str, zcode: int) -> int:
    url = "http://apis.data.go.kr/B552584/EvCharger/getChargerInfo"
    params = {
        "serviceKey": api_key,
        "numOfRows": 1,
        "pageNo": 1,
        "dataType": "JSON",
        "zcode": zcode
    }

    response = requests.get(url, params=params)
    if response.status_code != 200:
        print(f"❌ 요청 실패 (zcode={zcode}): {response.status_code}")
        return 0

    data = response.json()
    total_count = data.get("totalCount", 0)
    print(f"📊 총 {total_count}개의 충전소가 zcode={zcode} 지역에 존재합니다.")
    return total_count

def fetch_charger_data_by_region_with_checkpoint(
    api_key: str,
    output_dir: str = "Database/opendata/ev_charger_by_region",
    target_zcodes: list[int] = None,
    checkpoint_file: str = "charger_fetch_checkpoint.json"
) -> None:
    url = "http://apis.data.go.kr/B552584/EvCharger/getChargerInfo"
    os.makedirs(output_dir, exist_ok=True)

    checkpoint = {}
    if os.path.exists(checkpoint_file):
        with open(checkpoint_file, "r", encoding="utf-8") as f:
            checkpoint = json.load(f)

    target = target_zcodes if target_zcodes else list(ZCODE_DICT.keys())

    for zcode in target:
        region_name = ZCODE_DICT.get(zcode)
        if not region_name:
            print(f"⚠️ 알 수 없는 zcode: {zcode}")
            continue

        # 시작 페이지 설정
        start_page = 1
        if checkpoint.get("zcode") == zcode:
            start_page = checkpoint.get("page", 1)

        # 총 페이지 수 계산
        total_count = get_total_count_by_zcode(api_key, zcode)
        total_pages = (total_count // 1000) + 1
        print(f"📍 {region_name}({zcode}) 수집 시작 (page {start_page}부터, 총 {total_pages} 페이지 예상)\n")

        cleaned_name = normalize_region_name(region_name)
        filename = f"{cleaned_name}_ev_charger_info.csv"
        output_path = os.path.join(output_dir, filename)

        total_saved = 0
        if os.path.exists(output_path):
            existing_df = pd.read_csv(output_path, encoding="utf-8-sig")
            total_saved = len(existing_df)

        for page_no in tqdm(range(start_page, total_pages + 1), desc=f"{region_name} 진행 중", unit="page", initial=start_page-1, total=total_pages):
            params = {
                "serviceKey": api_key,
                "numOfRows": 1000,
                "pageNo": page_no,
                "dataType": "JSON",
                "zcode": zcode
            }

            response = requests.get(url, params=params)
            if response.status_code != 200:
                print(f"❌ 요청 실패 (zcode={zcode}, page={page_no}): {response.status_code}")
                break

            data = response.json()
            items = data.get("items", {}).get("item", [])
            if not items:
                print(f"✅ {region_name} 수집 완료 — 더 이상 데이터 없음 (총 {total_saved}건)")
                break

            df = pd.DataFrame(items)
            df["regionName"] = region_name

            df.to_csv(
                output_path,
                mode="a",
                index=False,
                encoding="utf-8-sig",
                header=not os.path.exists(output_path) or page_no == start_page
            )

            total_saved += len(df)

            checkpoint = {"zcode": zcode, "page": page_no + 1}
            with open(checkpoint_file, "w", encoding="utf-8") as f:
                json.dump(checkpoint, f, indent=2, ensure_ascii=False)

            time.sleep(0.3)

        print(f"\n📁 {region_name} 수집 완료: 총 {total_saved}건 저장 → {output_path}\n")
        if os.path.exists(checkpoint_file):
            os.remove(checkpoint_file)
def get_total_count_by_zcode(api_key: str, zcode: int) -> int:
    url = "http://apis.data.go.kr/B552584/EvCharger/getChargerInfo"
    params = {
        "serviceKey": api_key,
        "numOfRows": 1,
        "pageNo": 1,
        "dataType": "JSON",
        "zcode": zcode
    }

    response = requests.get(url, params=params)
    if response.status_code != 200:
        print(f"❌ 요청 실패 (zcode={zcode}): {response.status_code}")
        return 0

    data = response.json()
    total_count = data.get("totalCount", 0)
    print(f"📊 총 {total_count}개의 충전소가 zcode={zcode} 지역에 존재합니다.")
    return total_count
```

```python
import requests
import pandas as pd
import time
import os
import json

def fetch_gyeonggi_with_checkpoint(api_key: str, checkpoint_file="fetch_checkpoint.json"):
    url = "http://apis.data.go.kr/B552584/EvCharger/getChargerInfo"
    region_name = "경기도"
    zcode = 41
    output_dir = "Database/opendata/ev_charger_by_region"
    os.makedirs(output_dir, exist_ok=True)

    # ✅ 체크포인트 불러오기
    start_page = 1
    if os.path.exists(checkpoint_file):
        with open(checkpoint_file, "r", encoding="utf-8") as f:
            checkpoint = json.load(f)
            if checkpoint.get("zcode") == zcode:
                start_page = checkpoint.get("last_page", 1)

    print(f"📍 {region_name} 수집 시작 (page {start_page}부터)")

    region_data = pd.DataFrame()
    page_no = start_page

    while True:
        params = {
            "serviceKey": api_key,
            "numOfRows": 1000,
            "pageNo": page_no,
            "dataType": "JSON",
            "zcode": zcode
        }

        response = requests.get(url, params=params)
        if response.status_code != 200:
            print(f"❌ 요청 실패 (page={page_no}): {response.status_code}")
            break

        data = response.json()
        items = data.get("items", {}).get("item", [])
        if not items:
            print("✅ 더 이상 데이터 없음. 종료.")
            break

        df = pd.DataFrame(items)
        df["regionName"] = region_name
        region_data = pd.concat([region_data, df], ignore_index=True)

        # ✅ 체크포인트 저장
        progress_state = {
            "zcode": zcode,
            "region": region_name,
            "last_page": page_no + 1
        }
        with open(checkpoint_file, "w", encoding="utf-8") as f:
            json.dump(progress_state, f, indent=2, ensure_ascii=False)

        print(f"✅ page {page_no} 수집 완료 ({len(df)}건)")
        page_no += 1
        time.sleep(0.3)

    # ✅ 저장
    filename = f"{region_name}_ev_charger_info.csv"
    output_path = os.path.join(output_dir, filename)
    region_data.to_csv(output_path, index=False, encoding="utf-8-sig")
    print(f"📁 저장 완료 → {output_path}")
    print(f"📌 마지막 수집 페이지: {page_no - 1}")
```
