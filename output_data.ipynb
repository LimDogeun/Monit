```python
from geopy.distance import geodesic

def find_nearest_charger(lat, lng, df):
    """
    입력된 위도(lat), 경도(lng) 기준으로 가장 가까운 충전소 한 곳 반환
    """
    def distance(row):
        try:
            return geodesic((lat, lng), (row['lat'], row['lng'])).meters
        except:
            return float('inf')

    df['distance'] = df.apply(distance, axis=1)
    return df.sort_values('distance').iloc[0]
```

## 고속도로 방향 고려
- 경로 시간 찍어봤을때 기준으로 거르면은 바로 거를수 있을 것 같긴 함
